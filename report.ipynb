نویز فلفل نمکی تشکیل شده از پیکسل های سیاه(0) و سفید(255) می باشد. برای افزودن این نویز به تصویر، تابعی را تعریف می کنیم که بر اساس شدت نویز مورد نظر، در یک حلقه ی تکرار، تعدادی از پیکسل های تصویر را مشکی و تعدادی را سفید می کنیم. این تابع به صورت دستنویس نوشته شده است و در کد قابل مشاهده است.
</div>

```python
def sp_noise(img, rate = 0.1):
    h, w = img.shape[:2]
    density = h * w * rate
    for i in range(int(density/2)):
        ph = np.random.randint(1, h)
        pw = np.random.randint(1, w)
        img[ph, pw] = 0

    for i in range(int(density/2)):
        ph = np.random.randint(1, h)
        pw = np.random.randint(1, w)
        img[ph, pw] = 255

    return img
```

<div dir="rtl">
ساخت کرنل های laplacian و high-boost:

```python
def get_laplacian_filter(size):
    arr = []
    for j in range(size):
        arr.append([])
        for i in range(size):
            if j != int(size / 2) and i == int(size / 2):
                arr[j].append(-1)
            elif j == i and i == int(size / 2):
                arr[j].append(2*size - 2)
            elif j == int(size / 2) and i != int(size / 2):
                arr[j].append(-1)
            else:
                arr[j].append(0)

    return arr

def get_high_boost_filter(size, a):
    arr = []
    for j in range(size):
        arr.append([])
        for i in range(size):
            if i == j and i == int(size / 2):
                arr[j].append(a - 1/size**2)
            else:
                arr[j].append(-1/size**2)

    return arr
```

```python
def apply_sharpening_filter(img, filter_matrix, stream, cmap = None):
    img = img.copy()
    h, w = img.shape[:2]
    filter_matrix = np.array(filter_matrix)
    fh, fw = filter_matrix.shape

    # create output image
    out = np.array(img)

    for j in range(h - fh + 1):
        for i in range(w - fw + 1):
            acc = 0  # for RGB channels

            for fj in range(fh):
                for fi in range(fw):
                    acc += img[j + fj, i + fi] * filter_matrix[fj, fi]

            # clamp values
            acc = np.clip(acc, 0, 255)

            center_j = j + fh // 2
            center_i = i + fw // 2
            out[center_j, center_i] = acc.astype(np.uint8)

    if cmap is not None:
        plt.imsave(stream, out, cmap=cmap)
    else:
        plt.imsave(stream, out)

    return out
```

رفع‌تاری یا Deblurring مجموعه‌ای از روش‌ها و فیلترهای پردازش تصویر است که هدف آن‌ها بازگرداندن جزئیات از دست‌رفته در اثر تاری ناشی از حرکت، فوکوس نادرست یا محدودیت‌های اپتیکی است. در این مسئله، تصویر مشاهده‌شده معمولاً نتیجه‌ی یک تصویر اصلی است که توسط یک تابع پخش نقطه‌ای (Point Spread Function – PSF) با نویز ترکیب شده است. فیلترهای رفع‌تاری تلاش می‌کنند اثر PSF را تا جای ممکن معکوس کنند.<br>
1. فیلتر Inverse Filtering
ساده‌ترین روش رفع‌تاری است و بر پایه‌ی معکوس‌کردن مستقیم PSF عمل می‌کند.
مزیت: سرعت بالا و پیاده‌سازی بسیار ساده.
ضعف جدی: نسبت به نویز تقریباً کاملاً حساس است و در عمل زیاد قابل اعتماد نیست.
<br>
2. فیلتر Wiener Filtering
یکی از کلاسیک‌ترین و رایج‌ترین روش‌ها. علاوه بر PSF، مدل نویز و انرژی تصویر را نیز در نظر می‌گیرد.
مزیت: تعادل بین کاهش‌تاری و تقویت‌نشدن نویز، عملکرد بهتر از فیلتر معکوس.
ضعف: نیاز به تخمین نسبت سیگنال به نویز (SNR) و PSF.
<br>
3. روش‌های Regularized Inverse Filtering
در این روش‌ها معکوس‌کردن PSF با یک عبارت منظم‌سازی (Regularization) پایدار می‌شود. نمونه‌های رایج:
<br>
Tikhonov Regularization
<br>
Total Variation (TV) Deblurring
<br>
مزیت: جلوگیری از تقویت نویز و رفتار پایدارتر نسبت به روش معکوس ساده.
ضعف: نیاز به انتخاب پارامتر منظم‌سازی.
